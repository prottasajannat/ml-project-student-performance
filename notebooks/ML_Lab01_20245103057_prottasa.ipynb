```python
import random
import numpy as np
def generate_scores(use_seed):
    if use_seed:
      random.seed(42)
      np.random.seed(42)
    return np.random.randint(40, 101, size=5)
print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
print('The seed problem in machine learning occurs because random processes such as splitting data into training and testing sets,')
print('can produce different results each time. If we use a fixed random seed,such as 42, the same random results are generated every time,making the experiment reproducible.')
print('Without a seed,different runs may produce different results and accuracy scores.')
print('Therefore,using the same seed helps researchers compare models fairly and ensures that their results can be repeated.')
```

```python
## 1. Problem
Predict whether a student will pass or fail based on study hours, sleep hours, attendance, and snacks.
```

```python
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
display(df)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())
```

```python
baseline = df['passed'].mode()[0]
print("Baseline prediction:", baseline)
```

```python
from sklearn.model_selection import train_test_split

X = df[['study_hours', 'sleep_hours', 'attendance', 'snacks']]
y = df['passed']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```
